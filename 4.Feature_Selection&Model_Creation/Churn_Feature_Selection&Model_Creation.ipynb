## <span style="color:blue; font-size:22px;">4. Feature selection and Model Creation</span>

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
dataset = pd.read_csv("Churn_Modelling.csv")

df = dataset.copy()
```

```python
# Dropping the unnecessary Column
```

```python
df = df.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1)
```

```python
# Convert Categorical → Numeric
```

```python
df = pd.get_dummies(df, drop_first=True)
```

```python
# Define X and y
```

```python
X = df.drop('Exited', axis=1)
y = df['Exited']
```

```python
# Feature Selection
```

```python
num_features = 10  

model_fs = RandomForestClassifier(random_state=42)
model_fs.fit(X, y)

importances = pd.Series(model_fs.feature_importances_, index=X.columns)

selected_features = importances.nlargest(num_features).index

print("Selected Features:", list(selected_features))
```

```python
X_selected = X[selected_features]
```

```python
# Train-Test Split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_selected, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Defining the Models
```

```python
models = {
    "Logistic": LogisticRegression(max_iter=1000),
    "SVM Linear": SVC(kernel='linear'),
    "SVM RBF": SVC(kernel='rbf'),
    "KNN": KNeighborsClassifier(),
    "Naive Bayes": GaussianNB(),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42)
}
```

```python
# Training the Models and collecting the data
```

```python
results = []

for name, model in models.items():
    
    print(f"Training {name}...")
    
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    acc = accuracy_score(y_test, y_pred)
    
    results.append({
        "Model": name,
        "Accuracy": acc
    })
```

```python
# Result for 3
results_df = pd.DataFrame(results)

print("\nModel Performance Table:")
print(results_df)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import joblib

X_final = X   
y_final = y

# Split
X_train, X_test, y_train, y_test = train_test_split(
    X_final, y_final, test_size=0.2, random_state=42
)

# Scaling (important)
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Train final model
final_model = RandomForestClassifier(random_state=42)
final_model.fit(X_train, y_train)
joblib.dump(X.columns.tolist(), "features.pkl")
```
