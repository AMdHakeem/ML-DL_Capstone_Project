## <span style="color:blue; font-size:22px;">5. Deployment</span>

```python
# Loading the model, Scaler and Features
```

```python
import joblib
import pandas as pd

model = joblib.load("churn_model.pkl")
scaler = joblib.load("scaler.pkl")
feature_names = joblib.load("features.pkl")
```

```python
# Creating EMPTY DataFrame with correct structure
```

```python
new_data = pd.DataFrame(columns=feature_names)
```

```python
# Prediction coding
```

```python
new_data.loc[0] = 0  # initialize all values as 0

new_data['CreditScore'] = 600
new_data['Age'] = 40
new_data['Tenure'] = 5
new_data['Balance'] = 100000
new_data['NumOfProducts'] = 2
new_data['HasCrCard'] = 1
new_data['IsActiveMember'] = 1
new_data['EstimatedSalary'] = 50000
new_data['Gender_Male'] = 1
new_data['Geography_Germany'] = 0
new_data['Geography_Spain'] = 1
```

```python
# Scaling and Predict
```

```python
new_data_scaled = scaler.transform(new_data)

prediction = model.predict(new_data_scaled)

if prediction[0] == 1:
    print("Customer will EXIT ❌")
else:
    print("Customer will STAY ✅")
```
